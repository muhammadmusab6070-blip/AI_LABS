```python
import random
import matplotlib.pyplot as plt


# Simple Reflex Agent
def simple_reflex_agent(location, status):

    if status == "Dirty":
        return "Suck"

    elif location == "A" and status == "Clean":
        return "Right"

    elif location == "B" and status == "Clean":
        return "Left"


# Vacuum World Simulation
def simulate_vacuum_world(steps=10):

    # Create the world
    world = {
        "A": random.choice(["Dirty", "Clean"]),
        "B": random.choice(["Dirty", "Clean"])
    }

    # Starting location of agent
    location = random.choice(["A", "B"])

    # Lists for graph
    step_list = []
    action_list = []

    # Run simulation for 10 steps
    for step in range(1, steps + 1):

        # Get current status
        status = world[location]

        # Agent selects action
        action = simple_reflex_agent(location, status)

        # Print state
        print("Step:", step)
        print("Location:", location)
        print("Status:", status)
        print("Action:", action)
        print("World:", world)
        print("-------------------------")

        # Store step number
        step_list.append(step)

        # Convert actions into numbers for graph
        if action == "Suck":
            action_number = 0

        elif action == "Left":
            action_number = 1

        elif action == "Right":
            action_number = 2

        action_list.append(action_number)

        # Perform the action
        if action == "Suck":
            world[location] = "Clean"

        elif action == "Right":
            location = "B"

        elif action == "Left":
            location = "A"

    # Create graph
    plt.plot(step_list, action_list, marker="o")

    # Graph title and labels
    plt.title("Simple Reflex Vacuum Agent Simulation")
    plt.xlabel("Step")
    plt.ylabel("Action")

    # Change numbers into action names
    plt.yticks(
        [0, 1, 2],
        ["Suck", "Left", "Right"]
    )

    # Show graph
    plt.show()


# Run the simulation for 10 steps
simulate_vacuum_world(10)
```
